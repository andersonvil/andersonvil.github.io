```python
import json, statistics
from pathlib import Path

payload = json.loads(Path('synthetic_households.json').read_text(encoding='utf-8'))
assert payload['synthetic'] is True
groups = {}
for row in payload['records']:
    groups.setdefault(row['group'], []).append(row)
for group, rows in sorted(groups.items()):
    work = statistics.mean(row['market_work_hours'] for row in rows)
    care = statistics.mean(row['care_hours'] for row in rows)
    print(f'Group {group}: market work={work:.1f} hours; care={care:.1f} hours')
```
